```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Range of the x and y

x = np.linspace(-10,50,100)
```

```python
# Step function
step = np.where(x>=0,1,0)
plt.plot(x,step)
plt.title("Step activation function")
plt.xlabel('x')
plt.ylabel('y')
plt.grid(True)
plt.show()
```

```python
# sigmoid function

sigmoid = 1/(1+np.exp(-x))
plt.plot(x,sigmoid)
plt.title("signoid function")
plt.xlabel('x')
plt.ylabel('y')
plt.grid(True)
plt.show()
```

```python
# tanh
tanh = np.tanh(x)
plt.plot(x,tanh)
plt.title("tanh function")
plt.xlabel('x')
plt.ylabel('y')
plt.grid(True)
plt.show()
```

```python
# relu
relu = np.maximum(0,x)
plt.plot(x,relu)
plt.title("Step acticvation function")
plt.xlabel('x')
plt.ylabel('y')
plt.grid(True)
plt.show()
```
