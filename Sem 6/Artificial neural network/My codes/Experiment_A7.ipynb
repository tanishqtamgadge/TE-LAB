```python
# ===================================
# Artificial Neural Network (Forward + Backpropagation)
# ===================================
import numpy as np

# ===================================
# 1. Activation Function
# ===================================
def sigmoid(x):
    return 1 / (1 + np.exp(-x))

def sigmoid_derivative(x):
    return x * (1 - x)

# ===================================
# 2. Input (as given in manual)
# ===================================
X = np.array([[1, 0, 1]])
y = np.array([[1]])

# ===================================
# 3. Initialize weights and bias
# ===================================
np.random.seed(1)

w1 = np.random.rand(3, 3)   # input → hidden
w2 = np.random.rand(3, 1)   # hidden → output

b1 = np.zeros((1, 3))
b2 = np.zeros((1, 1))

# ===================================
# 4. Training
# ===================================
epochs = 3000
learning_rate = 0.1

for epoch in range(epochs):

    # ---------- Forward Propagation ----------
    z1 = np.dot(X, w1) + b1
    a1 = sigmoid(z1)

    z2 = np.dot(a1, w2) + b2
    a2 = sigmoid(z2)

    # ---------- Error ----------
    error = y - a2

    # ---------- Backpropagation ----------
    d2 = error * sigmoid_derivative(a2)

    d1 = d2.dot(w2.T) * sigmoid_derivative(a1)

    # ---------- Update weights ----------
    w2 += a1.T.dot(d2) * learning_rate
    w1 += X.T.dot(d1) * learning_rate

    # ---------- Update bias ----------
    b2 += np.sum(d2, axis=0, keepdims=True) * learning_rate
    b1 += np.sum(d1, axis=0, keepdims=True) * learning_rate

    # Print loss
    if epoch % 1000 == 0:
        print(f"Epoch {epoch}, Loss: {np.mean(np.abs(error)):.4f}")

# ===================================
# 5. Final Output
# ===================================
print("\nInput:", X)
print("Expected Output:", y)
print("Predicted Output:", a2)
```
