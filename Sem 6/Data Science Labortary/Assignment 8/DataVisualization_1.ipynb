```python
# Import required libraries
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Load dataset
dataset = sns.load_dataset('titanic')
```

```python
# Display first 5 rows
print(dataset.head())
```

```python
# 1. Distribution Plot (Age)

sns.histplot(dataset['age'], bins=10, kde=True)
plt.title("Age Distribution")
plt.show()
```

```python
# 2. Joint Plot (Age vs Fare)

sns.jointplot(x='age', y='fare', data=dataset, kind='scatter')
plt.show()
```

```python
# 3. Rug Plot (Fare)

sns.rugplot(dataset['fare'])
plt.title("Rug Plot of Fare")
plt.show()
```

```python
# 4. Bar Plot (Sex vs Age)
sns.barplot(x='sex', y='age', data=dataset)
plt.title("Average Age by Gender")
plt.show()
```

```python
# 5. Count Plot (Gender Count)

sns.countplot(x='sex', data=dataset)
plt.title("Count of Male and Female")
plt.show()
```

```python
# 6. Box Plot (Sex vs Age)
sns.boxplot(x='sex', y='age', data=dataset)
plt.title("Box Plot of Age by Gender")
plt.show()

```

```python
# 7. Violin Plot
sns.violinplot(x='sex', y='age', data=dataset)
plt.title("Violin Plot of Age Distribution")
plt.show()

```

```python
# 8. Strip Plot

sns.stripplot(x='sex', y='age', data=dataset, jitter=True)
plt.title("Strip Plot")
plt.show()
```

```python
# 9. Swarm Plot

sns.swarmplot(x='sex', y='age', data=dataset)
plt.title("Swarm Plot")
plt.show()

```

```python
# 10. Heatmap (Correlation)

corr = dataset.corr(numeric_only=True)
sns.heatmap(corr, annot=True)
plt.title("Correlation Heatmap")
plt.show()
```

```python
# 11. REQUIRED: Histogram of Fare

sns.histplot(dataset['fare'], bins=10, kde=False)
plt.title("Fare Distribution (Histogram)")
plt.xlabel("Fare")
plt.ylabel("Count")
plt.show()
```
