12 — Data Visualization

Histogram of Features

```python
plt.figure(figsize=(8,5))

sns.histplot(df['SepalLengthCm'], color='blue', label='Sepal Length')
sns.histplot(df['SepalWidthCm'], color='orange', label='Sepal Width')

plt.title("Distribution of Sepal Features")
plt.xlabel("Value")
plt.ylabel("Frequency")

plt.legend(title="Features")

plt.show()
```

Boxplot (Outlier Detection)

```python
plt.figure(figsize=(8,5))

sns.boxplot(data=df[['SepalLengthCm','SepalWidthCm','PetalLengthCm','PetalWidthCm']])

plt.title("Boxplot of Iris Features (Outlier Detection)")
plt.xlabel("Features")
plt.ylabel("Measurement (cm)")

plt.show()
```

Scatter PLot


```python
plt.figure(figsize=(8,5))

sns.scatterplot(
    data=df,
    x="SepalLengthCm",
    y="PetalLengthCm",
    hue="Species",
    palette="Set2"
)

plt.title("Sepal Length vs Petal Length by Species")

plt.legend(title="Species")

plt.show()
```

Pairplot (Feature Relationships)

```python
g = sns.pairplot(
    df,
    hue="Species",
    diag_kind="hist",
    palette="Set2"
)

# Add title
g.fig.suptitle("Pairwise Relationship of Iris Features", y=1.02)

# Show legend clearly
g._legend.set_title("Species")

plt.show()
```

13 — Correlation Analysis

```python
corr = df.corr(numeric_only=True)

plt.figure(figsize=(6,4))

sns.heatmap(corr, annot=True, cmap="coolwarm")

plt.title("Correlation Matrix of Iris Features")

plt.show()
```
